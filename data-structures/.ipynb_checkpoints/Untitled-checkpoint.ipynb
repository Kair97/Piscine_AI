```python
tags = ["python", "data", "ml", "python", "ai", "data", "python"]
```

```python
print(tags.count("python"))
print(tags.index("ml"))

if tags.count("python") > 1:
    f = True 
    res = []
    for tag in tags:
        if tag == "python":
            if f:
                res.append(tag)
                f = False
            else:
                continue 
        else:
            res.append(tag)
    print(res)
else:
    print(tags)
```

```python
# sort() modifies the original list
copy_prices = prices.copy()

print(copy_prices.sort())   # prints None
print(copy_prices)          # prints the sorted list

# sorted() creates a new list
print(sorted(prices))

# sort() returns None because it changes (mutates) the existing list in place.
# sorted() returns a new sorted list and leaves the original list unchanged.
```

## Exercise 4: Nested data structures

```python
students = [
    {"name": "Asel",   "scores": [88, 74, 95, 61]},
    {"name": "Dias",   "scores": [72, 85, 90, 78]},
    {"name": "Kamila", "scores": [91, 88, 77, 84]},
]
```

```python
for student in students:
    average = sum(student["scores"]) / len(student["scores"])
    print(f'{student["name"]}: {round(average, 2)}')
```

```python
hg = 0
name = ""
for student in students:
    average = sum(student["scores"]) / len(student["scores"])
    if average >= 85:
        student["grade"] = "A"
    elif average >= 70:
        student["grade"] = "B"
    else:
        student["grade"] = "C"
    if max(student["scores"]) > hg:
        name = student["name"]
        hg = max(student["scores"])
print(name)
print(students)
```

## Exercise 5: Comprehensions

```python
res = [i**2 for i in range(1, 16) if i % 2 != 0]
print(res)
```

```python
words = ["Hello", "  world  ", "Python", " data ", "AI"]
```

```python
result = []

for word in words:

    clean_word = word.strip()

    if len(clean_word) > 2:
        result.append(clean_word.lower())

print(result)
```

```python
eres = {}
for p, c in inventory.items():
    if c > 0: 
        eres[p] = "in stock"
    else:
        eres[p] = "out of stock"
print(eres)
```

```python
a = [1, 2, 3, 4, 5]
b = [4, 5, 6, 7, 8]
```

```python
ib = {num for num in a if num in b}
print(ib)
```

## Exercise 6: Built-in functions on collections

```python
numbers = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]
```

```python
narr = sorted(numbers, key=lambda x: (x%3, x))
print(narr)
```

```python
swithga = list(filter(lambda x: x["grade"] == "A", students))
for s in swithga:
    print(s["name"])
```

```python
ms = map(lambda x: f"Name: {x["name"]}, Grade: {x["grade"]}", students)
for m in ms:
    print(m)
```

```python
v1 = [1, 2, 3]
v2 = [4, 5, 6]
```

```python
ps = [pair for pair in zip(v1, v2)]
print(ps)
dot =  sum(map(lambda x: x[0] * x[1], ps))
print(dot)
```
