```python
def apply_twice(f, x):
    ft = f(x) 
    return f(ft)

print(apply_twice(lambda x: x*2, 3))
```

```python
words = ["elephant", "cat", "sky", "programming", "ai", "rhythm"]

vowels = "aeiouy"
def cv(str):
    return sum(1 for x in str.lower() if x in vowels)

sorted_words = sorted(words, key=lambda x: (cv(x), len(x)))
print(sorted_words)
```

```python
import functools
import time 

def timer(f):
    @functools.wraps(f)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = f(*args, **kwargs)
        end_time = time.perf_counter()
        duration_ms = (end_time - start_time) * 1000
        print(f"{f.__name__} took {duration_ms:.2f} ms")
        return result
    return wrapper

@timer
def slow_sum(n):
    """Computes sum of a large range."""
    return sum(range(n))

print(slow_sum(10_000_000))
print(f"Preserved Name: {slow_sum.__name__}")
```

```python
def retry(max_attempts=3):
    def decorator(f):
        @functools.wraps(f)
        def wrapper(*args, **kwargs):
            last_exception = None
            for attempt in range(max_attempts):
                try:
                    return f(*args, **kwargs)
                except Exception as e:
                    last_exception = e
            raise last_exception
        return wrapper
    return decorator
call_count = 0

@retry(max_attempts=3)
def flaky_connection():
    global call_count
    call_count += 1
    if call_count < 3:
        raise ConnectionError("timeout")
    return "connected"

print(flaky_connection())
print(f"Total attempts: {call_count}")
```

```python
import itertools

def fibonacci():
    a, b = 0, 1 
    while True:
        yield a 
        a, b = b, a+b
```

```python
first_12 = itertools.islice(fibonacci(), 12)
print(*(first_12))
```

```python
def read_in_chunks(filepath, chunk_size):
    with open("haiku.txt", 'r') as f:
        chunk = []
        for line in f:
            chunk.append(line)
            if len(chunk) == chunk_size:
                yield chunk
                chunk = []
        if chunk:
            yield chunk
for chunk in read_in_chunks("haiku.txt", 2):
    print(chunk)
```

```python
transactions = [
    {"category": "food",    "amount": 12.5},
    {"category": "food",    "amount": 8.0},
    {"category": "travel",  "amount": 220.0},
    {"category": "travel",  "amount": 45.0},
    {"category": "utilities","amount": 60.0},
]

for cat, group in itertools.groupby(transactions, key=lambda x: x["category"]):
    total = sum(i["amount"] for i in group)
    print(f"{cat}: {total}")
```

```python
import functools
import time 

def fib_without_cache(n):
    if n < 2:
        return n 
    return fib_without_cache(n-1) + fib_without_cache(n-2)

@functools.lru_cache(maxsize=None)
def fib_with_cache(n):
    if n < 2:
        return n
    return fib_with_cache(n-1) + fib_with_cache(n-2)

start_raw = time.perf_counter()
fib_without_cache(35)
end_raw = time.perf_counter()
time_without_ms = (end_raw - start_raw) * 1000

fib_with_cache.cache_clear()

start_cached = time.perf_counter()
fib_with_cache(35)
end_cached = time.perf_counter()
time_with_ms = (end_cached - start_cached) * 1000

speedup = time_without_ms / time_with_ms

print(f"Without cache: {time_without_ms:.2f} ms")
print(f"With cache:    {time_with_ms:.2f} ms")
print(f"Speedup:       {int(speedup)}x")
```

```python
nums = [1, 2, 3, 4, 5]

def power(b, e):
    return b**e

square = functools.partial(power, e=2)
cube = functools.partial(power, e=3)

print(list(map(square, nums)))
print(list(map(cube, nums)))
```
