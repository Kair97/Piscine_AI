## Exercise 4: Control flow

```python
def classify_bmi(bmi):
    if bmi < 18.5:
        print("Underweight") 
    elif bmi < 25: 
        print("Normal")
    elif bmi < 30:
        print("Overweight")
    else:
        print("Obese")
```

```python
nums = [16.0, 22.4, 27.1, 35.8]
for num in nums:
    classify_bmi(num)
```

```python
def fizzbuzz(n):
    if n % 3 == 0 and n % 5 == 0: 
        return "FizzBuzz"
    elif n % 3 == 0 :
        return "Fizz"
    elif n % 5 == 0:
        return "Buzz"
    else:
        return str(n)
```

```python
for i in range(1, 21):
    print(fizzbuzz(i), end=" ")
```

## Exercise 6: Functions

```python
def temperature_convert(value, unit="C"):
    if unit == "C":
        return value * 9/5 + 32
    else:
        return (value - 32) * 5/9
```

```python
print(temperature_convert(100))      ##→ 212.0
print(temperature_convert(32, "F"))  ##→ 0.0
print(temperature_convert(-40))      ##→ -40.0
```

```python
def summarize(*values):
    if not values:
        return {"count": 0, "sum": 0, "min": None, "max": None, "mean": 0}
        
    total_count = len(values)
    total_sum = sum(values)
    minimum_val = min(values)
    maximum_val = max(values)
    mean_val = total_sum / total_count
    
    return {
        "count": total_count,
        "sum": total_sum,
        "min": minimum_val,
        "max": maximum_val,
        "mean": mean_val
    }
    
```

```python
print(summarize(4, 7, 2, 9, 1))
```

```python
def format_record(**fields):
    pairs = [f"{key}={value}" for key, value in sorted(fields.items())]

    return " | ".join(pairs)
```

```python
print(format_record(city="Almaty", country="Kazakhstan", population=2000000))
```

## Exercise 7: Scope

```python
x = "global"

def outer():
    x = "enclosing"
    def inner():
        x = "local"
        print(x)
    inner()
    print(x)

outer()
print(x)
```

```python
counter = 0
def increment_counter():
    global counter 
    counter += 1
    print(counter)
```

```python
increment_counter()
increment_counter()
increment_counter()
```

```python
def make_multiplier(factor):
    def multiply(val):
        return val * factor
        
    return multiply
```

```python
double = make_multiplier(2)
triple = make_multiplier(3)
print(double(5))
print(triple(5))
```
