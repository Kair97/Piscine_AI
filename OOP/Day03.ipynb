## Exercise 1: Your first class

```python
class Rectangle: 
    count = 0
    def __init__(self, width, height):
        self.width = float(width)
        self.height = float(height)
        Rectangle.count += 1 

    def area(self):
        return self.width * self.height
    def perimeter(self):
        return 2*(self.width+self.height)
    def is_square(self):
        return self.width == self.height
```

```python
r1 = Rectangle(4, 6)
r2 = Rectangle(5, 5)
print(f"Area: {r1.area()} | Perimeter: {r1.perimeter()} | Square: {r1.is_square()}")
print(f"Area: {r2.area()} | Perimeter: {r2.perimeter()} | Square: {r2.is_square()}")
print(Rectangle.count)
```

```python
r1.width = -50
print(r1.area())
```

```python
print(r2.count)
```

## Exercise 2: Encapsulation and properties

```python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius
    @property 
    def celsius(self):
        return self._celsius
    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Temperature cannot be below absolute zero (-273.15°C)")
        else:
            self._celsius = value 

    @property
    def fahrenheit(self):             
        return self._celsius * 9/5 + 32

    @fahrenheit.setter
    def fahrenheit(self, value):
        self.celsius = (value - 32) * 5/9
```

```python
t = Temperature(25)
print(t.celsius)
print(t.fahrenheit)
t.fahrenheit = 32
print(t.celsius)
```

```python
try:
    t.celsius = -300
except ValueError as e:
    print(e)
```

## Exercise 3: Inheritance

```python
class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species
    def describe(self):
        return f"{self.name} is a {self.species}"
    def speak(self):
        return "..."

class Dog(Animal):
    def __init__(self, name):
        super().__init__(name, species="Canis lupus familiaris")
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def __init__(self, name):
        super().__init__(name, species="Felis catus")
    def speak(self):
        return "Meow!"
```

```python
rex = Dog("Rex")
whiskers = Cat("Whiskers")

print(rex.describe())
print(rex.speak())
print(whiskers.describe())
print(whiskers.speak())
```

```python
print(isinstance(rex, Dog))
print(isinstance(rex, Animal))
print(issubclass(Dog, Animal))  
```

## Exercise 4: Dunder methods


```python
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __repr__(self):
        return f"Vector2D({self.x}, {self.y})"
    def __str__(self):
        return f"({self.x}, {self.y})"
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
    def __add__(self, other):
        return Vector2D(self.x + other.x, self.y + other.y)
    def __len__(self):
        return 2
```

```python
v1 = Vector2D(1, 2)
v2 = Vector2D(3, 4)
v3 = Vector2D(1, 2)
print(repr(v1))
print(str(v1))
print(v1 == v3)
print(v1 == v2)
print(v1 + v2)
print(len(v1))
```

## Exercise 5: Class methods and static methods

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius**2
    @classmethod 
    def from_diameter(cls, diameter):
        return cls(diameter/2)
    @staticmethod
    def is_valid_radius(radius):
        return radius > 0 
```

```python
import math
c1 = Circle(5)
c2 = Circle.from_diameter(10)
print(round(c1.area(), 4))
print(round(c2.area(), 4))
print(Circle.is_valid_radius(3))
print(Circle.is_valid_radius(-1))
```

 Use @classmethod when the method needs the class itself — usually as an
 alternate constructor that builds and returns a new instance via cls(...).
 Use @staticmethod when the method needs neither the instance nor the class;
 it's just a related helper function grouped inside the class for organization.

## Exercise 6: Dataclasses

```python
from dataclasses import dataclass

@dataclass(frozen=True)
class SensorReading:
    sensor_id: str
    timestamp: float
    value: float 
    unit: str = "°C"
```

```python
r1 = SensorReading("TEMP_01", 1700000000.0, 23.5)
r2 = SensorReading("TEMP_01", 1700000000.0, 23.5)
r3 = SensorReading("TEMP_02", 1700000001.0, 24.1, unit="°F")
print(r1)
print(r1 == r2)
print(r1 == r3)
```

```python
r1 = SensorReading("TEMP_01", 1700000000.0, 23.5)

try:
    r1.value = 99.0
except Exception as e:
    print(type(e))  
```
